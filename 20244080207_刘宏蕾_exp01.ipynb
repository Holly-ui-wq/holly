```python
# ==========================================
# 任务 1：分数等级判断函数（条件与异常处理）
# ==========================================
def score_level(score):
    # 检查分数范围，超出 0-100 抛出 ValueError
    if score < 0 or score > 100:
        raise ValueError("分数必须在 0 到 100 之间！")
    
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

print("--- 任务 1：测试 score_level ---")
print(f"95分: {score_level(95)}")
print(f"85分: {score_level(85)}")
print(f"72分: {score_level(72)}")
print(f"50分: {score_level(50)}")
try:
    score_level(105)
except ValueError as e:
    print(f"捕获到异常: {e}")


# ==========================================
# 任务 2：for 和 while 计算 1 到 100 的和
# ==========================================
print("\n--- 任务 2：计算 1-100 的和 ---")

# 使用 for 循环
sum_for = 0
for i in range(1, 101):  # range(1, 101) 包含 1 到 100
    sum_for += i
print(f"for 循环结果: {sum_for}")

# 使用 while 循环
sum_while = 0
j = 1
while j <= 100:
    sum_while += j
    j += 1
print(f"while 循环结果: {sum_while}")

print(f"两种结果是否一致？ {sum_for == sum_while}")


# ==========================================
# 任务 3：递归版和迭代版阶乘
# ==========================================
print("\n--- 任务 3：递归 vs 迭代求阶乘 ---")

# 递归版
def factorial_recursive(n):
    if n == 0:  # 约定 0! = 1
        return 1
    else:
        return n * factorial_recursive(n - 1)

# 迭代版
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# 比较测试
for n in [0, 1, 5, 10]:
    rec_res = factorial_recursive(n)
    iter_res = factorial_iterative(n)
    print(f"n={n}: 递归结果={rec_res}, 迭代结果={iter_res}, 是否相同={rec_res == iter_res}")


# ==========================================
# 任务 4：summary 函数返回元组与解包
# ==========================================
print("\n--- 任务 4：summary 函数 ---")

def summary(values):
    # 题目说明是非空列表，直接使用内置函数
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)  # 返回一个元组

test_list = [10, 20, 5, 40, 15]
# 通过元组解包接收结果
minimum, maximum, average = summary(test_list)
print(f"数据列表: {test_list}")
print(f"最小值: {minimum}, 最大值: {maximum}, 平均值: {average}")


# ==========================================
# 任务 5：可变参数 *args
# ==========================================
print("\n--- 任务 5：sum_all 可变参数 ---")

def sum_all(*args):
    # args 是一个元组，包含所有传入的参数
    return sum(args)  # 也可以自己用 for 循环写

print(f"无参数: {sum_all()}")
print(f"单个参数: {sum_all(10)}")
print(f"多个参数: {sum_all(1, 2, 3, 4, 5)}")
```

```python
# 初始数据
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

print("=========== 任务 1：zip 构建字典 ===========")
# 使用 zip 将两个列表组合，并用 dict() 转换为字典
score_dict = dict(zip(names, scores))
# 使用 items() 逐项输出
for name, score in score_dict.items():
    print(f"姓名: {name}, 成绩: {score}")


print("\n=========== 任务 2：enumerate 输出序号 ===========")
# enumerate 的 start 参数可以指定起始序号为 1
for index, (name, score) in enumerate(score_dict.items(), start=1):
    print(f"序号: {index}, 姓名: {name}, 成绩: {score}")


print("\n=========== 任务 3：列表推导式 ===========")
# 1. 提取及格（>=60）学生的姓名
passed_names = [name for name, score in zip(names, scores) if score >= 60]
print(f"及格学生名单: {passed_names}")

# 2. 生成所有成绩的平方列表
squared_scores = [score ** 2 for score in scores]
print(f"成绩的平方列表: {squared_scores}")


print("\n=========== 任务 4：基础统计与格式化 ===========")
count = len(scores)
avg_score = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
# 及格率：及格人数 / 总人数 * 100
pass_rate = (len(passed_names) / count) * 100 

# 字符串格式化：平均分保留两位小数，及格率加上百分号
print(f"总人数: {count} 人")
print(f"平均分: {avg_score:.2f} 分")
print(f"最高分: {max_score} 分")
print(f"最低分: {min_score} 分")
print(f"及格率: {pass_rate:.2f}%")


print("\n=========== 任务 5：别名 (alias) 与 副本 (copy) ===========")
# 使用单独的演示列表，避免破坏原数据
demo_list = [10, 20, 30]

# alias 是引用（别名），指向同一个内存地址
alias = demo_list
# copy() 是浅拷贝，创建了一个新的对象
copied = demo_list.copy()

# 修改 alias 中的第一个元素
alias[0] = 999

print(f"原始列表 demo_list: {demo_list}")
print(f"别名 alias: {alias}")
print(f"副本 copied: {copied}")
# 结论：修改 alias 会导致原列表跟着变，因为它们指向同一个对象；而 copy 不受影响。


print("\n=========== 任务 6：元组与不等长 zip ===========")
# 1. 元组保存单条记录，验证不可变
record = ("Alice", 85)
print(f"原始元组记录: {record}")
try:
    record[0] = "Bob"  # 尝试修改元组的第一个元素
except TypeError as e:
    print(f"捕获报错: {e}")
    print("原因: 元组是不可变数据类型，创建后不能修改元素。")

# 2. 长度不同的两个序列使用 zip()
list_a = [1, 2, 3]
list_b = ['a', 'b']
# zip 会以长度最短的序列为准，多出来的元素会被直接丢弃
zipped_unequal = list(zip(list_a, list_b))
print(f"不等长序列 zip 的结果: {zipped_unequal}")
# 结果只有两个元组，list_a 中的 3 被忽略了。
```

```python
%%writefile student_utils.py
# 这是 student_utils.py 文件的内容

def parse_student(line):
    """
    解析单行学生记录。
    格式："姓名,成绩"
    """
    # 1. 去除首尾空白并按照逗号分割
    parts = line.strip().split(",")
    
    # 2. 检查字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，预期为 2，实际为 {len(parts)}")
        
    name = parts[0].strip()
    score_str = parts[1].strip()
    
    # 3. 检查姓名是否为空
    if not name:
        raise ValueError("姓名字段为空")
        
    # 4. 将成绩转换为整数，若失败则抛出异常
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩 '{score_str}' 不是有效的整数")
        
    # 5. 检查成绩范围
    if not (0 <= score <= 100):
        raise ValueError(f"成绩 {score} 超出 0~100 范围")
        
    # 有效记录返回字典
    return {"name": name, "score": score}
```

```python
import copy

print("=========== 任务 1：定义 Student 类 ===========")
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        # 成绩大于等于 60 分及格
        return self.score >= 60

    def __str__(self):
        # 自定义打印格式
        status = "及格" if self.is_passed() else "不及格"
        return f"学生: {self.name}, 成绩: {self.score}, 状态: {status}"


print("\n=========== 任务 2：为有效记录创建对象 ===========")
# 实验4中的有效记录
valid_records = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92},
    {"name": "David", "score": 76},
    {"name": "Eva", "score": 60},
    {"name": "Frank", "score": 89}
]

# 遍历记录创建对象并输出
for record in valid_records:
    student_obj = Student(record["name"], record["score"])
    # 触发 __str__ 方法
    print(student_obj) 


print("\n=========== 任务 3：定义 GraduateStudent 子类 ===========")
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 使用 super() 调用父类的初始化方法
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        # 可以复用父类的 __str__ 方法，并加上自己的属性
        return f"{super().__str__()}, 研究方向: {self.research_topic}"

# 测试子类
grad_stu = GraduateStudent("Alice", 85, "人工智能")
print(grad_stu)  # 输出: 学生: Alice, 成绩: 85, 状态: 及格, 研究方向: 人工智能


print("\n=========== 任务 4：对象复制（浅拷贝 vs 深拷贝） ===========")

# 定义一个生成新原始对象的辅助函数，保证每次比较都是独立的
def create_original():
    return {"name": "Alice", "scores": [85, 90]}

# --- 1. 直接赋值 (Alias) ---
print("\n--- 测试 1：直接赋值 ---")
orig_1 = create_original()
alias = orig_1
# 修改内部列表
alias["scores"].append(100)
print(f"原始对象: {orig_1}")
print(f"别名对象: {alias}")
# 结论：直接赋值仅仅是贴了个标签，指向同一个内存地址。修改 alias 会连带原对象一起改变。

# --- 2. 浅拷贝 (Shallow Copy) ---
print("\n--- 测试 2：浅拷贝 copy.copy() ---")
orig_2 = create_original()
shallow_copy = copy.copy(orig_2)
# 修改内部列表
shallow_copy["scores"].append(100)
print(f"原始对象: {orig_2}")
print(f"浅拷贝对象: {shallow_copy}")
# 结论：浅拷贝只复制了最外层（字典本身），但内部嵌套的列表依然是共用的。修改内部列表，原对象也会变。

# --- 3. 深拷贝 (Deep Copy) ---
print("\n--- 测试 3：深拷贝 copy.deepcopy() ---")
orig_3 = create_original()
deep_copy = copy.deepcopy(orig_3)
# 修改内部列表
deep_copy["scores"].append(100)
print(f"原始对象: {orig_3}")
print(f"深拷贝对象: {deep_copy}")
# 结论：深拷贝连内部嵌套的对象（列表）也一并克隆了一份。修改新对象，原对象完全不受影响。
```
