```python
# 1. 定义 say_hi 函数
def say_hi():
    print("Hello, world!")

# 2. 定义 goat_pad 函数
def goat_pad(text):
    # 使用 + 号将字符串拼接起来
    print("goat" + text + "goat")

# 3. 定义 print_n 函数
def print_n(s, n):
    # 使用 for 循环，重复执行 n 次
    for i in range(n):
        print(s)

# ==========================================
# 下面是测试代码，用来验证函数是否写对了
# ==========================================
print("--- 测试 1 ---")
say_hi()

print("\n--- 测试 2 ---")
goat_pad("apple")  # 传入 "apple" 看看会打印什么

print("\n--- 测试 3 ---")
print_n("Python", 3)  # 传入 "Python" 和 3，它应该会打印 3 行
```

```python
# 1. 定义 gcd 函数
def gcd(m, n):
    while n != 0:
        # 将 n 赋值给 m，将 m 除以 n 的余数赋值给 n
        m, n = n, m % n
    return m
```

```python
# 2. 测试数字对
print("(2018, 2019) 的 GCD 是:", gcd(2018, 2019))
print("(1200, 300) 的 GCD 是:", gcd(1200, 300))
print("(5040, 60) 的 GCD 是:", gcd(5040, 60))
```

```python
import math

# 1. 使用极限公式 (2) 近似 e
def euler_limit(n):
    # 公式：(1 + 1/x)^x，这里把 x 替换为参数 n
    return (1 + 1 / n) ** n

# 2. 使用无限求和公式 (1) 的前 n 项近似 e
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    
    total = 0.0
    # 因为公式是从 k=0 开始，前 n 项即 k 从 0 到 n-1
    for k in range(n):
        # math.factorial(k) 计算 k!
        total += 1 / math.factorial(k)
    return total

# 3. 使用 while 循环逼近 e，直到误差在 epsilon 以内
def euler_approx(epsilon):
    true_e = math.exp(1) # 获取 e 的真实值
    total = 0.0
    k = 0
    
    # 当当前累加值与真实值的差值大于等于 epsilon 时，继续循环
    while abs(true_e - total) >= epsilon:
        total += 1 / math.factorial(k)
        k += 1
        
    return total

# 4. 打印从 1 到 n 的每一项的值
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        # 获取前 k 项的近似值
        approx_value = euler_infinite_sum(k)
        print(f"n={k} 时: {approx_value}")
```

```python
def is_power_of_2(n):
    # 1 是 2 的 0 次方
    if n == 1:
        return True
    
    # 如果是小于 1 的数，按照题目假设不会输入，但为了严谨可以加上
    if n <= 0:
        return False
        
    # 只要 n 大于 1，就不断除以 2
    while n > 1:
        if n % 2 != 0: # 如果不能被 2 整除，说明有其他的质因数，不是 2 的幂
            return False
        n = n // 2 # 使用整除（向下取整）保持 n 是整数
        
    return True
```

```python
def is_power(b, n):
    # 任何正整数的 0 次方都是 1
    if n == 1:
        return True
        
    # 如果底数 b 是 1，只有 n 也是 1 的时候才成立（已经在上一步返回了）
    # 如果 n 不是 1 且 b 是 1，那 n 永远不可能是 1 的幂
    if b == 1:
        return False
        
    # 循环判断
    while n > 1:
        if n % b != 0: # 如果不能被底数 b 整除，则不是幂
            return False
        n = n // b # 继续除以底数 b
        
    return True
```

```python
print("--- 测试 is_power_of_2 ---")
print(f"1 是 2 的幂吗? {is_power_of_2(1)}")      # True (2^0)
print(f"8 是 2 的幂吗? {is_power_of_2(8)}")      # True (2^3)
print(f"10 是 2 的幂吗? {is_power_of_2(10)}")    # False
print(f"64 是 2 的幂吗? {is_power_of_2(64)}")    # True

print("\n--- 测试 is_power ---")
print(f"9 是 3 的幂吗? {is_power(3, 9)}")        # True (3^2)
print(f"10 是 3 的幂吗? {is_power(3, 10)}")      # False
print(f"125 是 5 的幂吗? {is_power(5, 125)}")    # True (5^3)
print(f"1 是 7 的幂吗? {is_power(7, 1)}")        # True (7^0)
```
