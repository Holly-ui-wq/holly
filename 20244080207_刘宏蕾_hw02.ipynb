```python
def is_palindrome(s):
    # 转换为小写并去除空格
    cleaned_s = s.lower().replace(" ", "")
    # 判断正序和倒序是否一致
    return cleaned_s == cleaned_s[::-1]

# 测试
print(is_palindrome("tacocat"))        # True
print(is_palindrome("T A C O cat"))    # True
print(is_palindrome("rats live on no evil star")) # True
print(is_palindrome("hello"))          # False
```

```python
def is_abecedarian(s):
    # 转换为小写并去除空格
    cleaned_s = s.lower().replace(" ", "")
    # 比较原字符串列表与排序后的列表
    return list(cleaned_s) == sorted(cleaned_s)

# 测试
print(is_abecedarian("adder"))       # True (a <= d <= d <= e <= r)
print(is_abecedarian("beet"))        # True
print(is_abecedarian("abcd efgh xyz")) # True
print(is_abecedarian("dog"))         # False
print(is_abecedarian("cat"))         # False
```

```python
def remove_vowels(s):
    # 定义元音字母（包含大小写以防万一）
    vowels = "aeiouAEIOU"
    # 过滤掉元音字母
    return "".join([char for char in s if char not in vowels])

# 测试
print(remove_vowels("cat"))     # 'ct'
print(remove_vowels("audio"))   # 'd'
print(remove_vowels("aeiou"))   # '' (全部由元音组成返回空字符串)
print(remove_vowels("Hello World")) # 'Hll Wrld'
```

```python
def list_reverse(lst):
    # 检查输入是否为列表
    if not isinstance(lst, list):
        raise TypeError("输入必须是一个列表 (Input must be a list)")
    
    # 使用切片语法返回反转后的新列表
    return lst[::-1]

# 测试用例
print(list_reverse([1, 2, 3]))  # 输出: [3, 2, 1]
try:
    list_reverse("hello")
except TypeError as e:
    print(f"捕获到错误: {e}")   # 输出: 捕获到错误: 输入必须是一个列表 (Input must be a list)
```

```python
def is_sorted(seq):
    # 遍历序列，直到倒数第二个元素
    for i in range(len(seq) - 1):
        # 如果当前元素大于下一个元素，说明不是升序
        if seq[i] > seq[i + 1]:
            return False
    # 如果遍历完都没有发现逆序，说明是升序排列
    return True

# 测试用例
print(is_sorted([1, 2, 2, 3]))     # 输出: True (允许相等)
print(is_sorted([1, 3, 2, 4]))     # 输出: False
print(is_sorted("abcde"))          # 输出: True (字符串也是序列)
print(is_sorted([10, 5, 1]))       # 输出: False
print(is_sorted([42]))             # 输出: True (单元素序列视为已排序)
print(is_sorted([]))               # 输出: True (空序列视为已排序)
```

```python
# ================= 函数定义部分 =================

def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("参数 t 必须是一个元组")
    if not isinstance(s, (int, float)):
        raise TypeError("参数 s 必须是整数或浮点数")
    for val in t:
        if not isinstance(val, (int, float)):
            raise TypeError("元组 t 内的元素必须是整数或浮点数")
    return tuple(val * s for val in t)

def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("输入的两个参数必须都是元组")
    if len(x) != len(y):
        raise ValueError("两个元组的长度必须一致")
    for val in x + y:
        if not isinstance(val, (int, float)):
            raise TypeError("元组内的元素必须是整数或浮点数")
    return sum(a * b for a, b in zip(x, y))

def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    if not isinstance(mx[0], tuple):
        return False
    row_length = len(mx[0])
    for row in mx:
        if not isinstance(row, tuple):
            return False
        if len(row) != row_length:
            return False
        for val in row:
            if not isinstance(val, (int, float)):
                return False
    return True

def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("输入的矩阵不合法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是一个元组")
    for val in vec:
        if not isinstance(val, (int, float)):
            raise TypeError("向量内的元素必须是整数或浮点数")
    if len(mx) == 0:
        return ()
    if len(mx[0]) != len(vec):
        raise ValueError("矩阵的列数必须等于向量的长度")
    result = []
    for row in mx:
        dot_product = sum(r_val * v_val for r_val, v_val in zip(row, vec))
        result.append(dot_product)
    return tuple(result)


# ================= 测试代码部分 =================

# 测试 1
print("测试 1:", vec_scalar_mult((1, 2, 3), 2.5))  # 预期: (2.5, 5.0, 7.5)

# 测试 2
print("测试 2:", vec_inner_product((1, 2, 3), (4, 5, 6)))  # 预期: 32

# 测试 3
valid_mx = ((1, 2), (3, 4), (5, 6))
invalid_mx = ((1, 2), (3, 4, 5))
print("测试 3-1:", check_valid_mx(valid_mx))    # 预期: True
print("测试 3-2:", check_valid_mx(invalid_mx))  # 预期: False

# 测试 4
matrix = ((1, 2), (3, 4), (5, 6))
vector = (1, 1)
print("测试 4:", mx_vec_mult(matrix, vector))  # 预期: (3, 7, 11)
```

```python
class Vector:
    def __init__(self, dim, elements=None):
        # 1. 验证维度类型和合法性
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
            
        self.dim = dim
        
        # 处理默认行为：如果未提供元素，默认全为 0
        if elements is None:
            self.elements = tuple(0 for _ in range(dim))
        else:
            # 验证元素类型
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或元组")
            # 验证长度一致性
            if len(elements) != dim:
                raise ValueError("维度与元素个数不一致")
            # 验证元素内部数据是否为数字
            for el in elements:
                if not isinstance(el, (int, float)):
                    raise TypeError("元素内部必须是整数或浮点数")
            # 使用元组存储，保证对象状态不可变
            self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # ================= 问题 3：比较运算符 =================
    def _check_comparable(self, other):
        """内部辅助方法：检查类型和维度是否一致"""
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("无法比较维度不同的两个向量")
        return True

    def __eq__(self, other):
        if self._check_comparable(other) == NotImplemented:
            return False
        return self.elements == other.elements

    def __lt__(self, other):
        if self._check_comparable(other) == NotImplemented:
            return NotImplemented
        # Python 的元组自带字典序比较，直接复用
        return self.elements < other.elements

    def __le__(self, other):
        if self._check_comparable(other) == NotImplemented:
            return NotImplemented
        return self.elements <= other.elements

    def __gt__(self, other):
        if self._check_comparable(other) == NotImplemented:
            return NotImplemented
        return self.elements > other.elements

    def __ge__(self, other):
        if self._check_comparable(other) == NotImplemented:
            return NotImplemented
        return self.elements >= other.elements

    # ================= 问题 4：点积方法 =================
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("传入的参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("计算内积时，两个向量的维度必须一致")
        
        return sum(a * b for a, b in zip(self.elements, other.elements))

    # ================= 问题 5：乘法运算符 =================
    def __mul__(self, other):
        # 情况 A：向量乘以标量（v * 2）
        if isinstance(other, (int, float)):
            new_elements = tuple(el * other for el in self.elements)
            return Vector(self.dim, new_elements)
            
        # 情况 B：向量乘以向量，即逐元素相乘（v * w）
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("逐元素相乘时，两个向量的维度必须一致")
            new_elements = tuple(a * b for a, b in zip(self.elements, other.elements))
            return Vector(self.dim, new_elements)
            
        else:
            raise TypeError("不支持的操作数类型")

    def __rmul__(self, other):
        # 标量左乘向量（2 * v），逻辑同 __mul__
        if isinstance(other, (int, float)):
            return self.__mul__(other)
        return NotImplemented

# ================= 测试代码 =================

if __name__ == "__main__":
    v1 = Vector(3, (1, 2, 3))
    v2 = Vector(3, (4, 5, 6))
    v3 = Vector(2, (2, 4))
    v4 = Vector(2, (2, 5))
    
    # 测试问题 1 默认行为
    print("默认初始化:", Vector(2)) # Vector(2, (0, 0))
    
    # 测试问题 3 比较运算
    print("相等测试:", v1 == Vector(3, [1, 2, 3])) # True
    print("字典序比较:", v3 < v4) # True (2,4) < (2,5)
    
    # 测试问题 4 内积
    print("内积:", v1.dot(v2)) # 1*4 + 2*5 + 3*6 = 32
    
    # 测试问题 5 乘法
    print("标量乘法 (v1 * 2):", v1 * 2)      # Vector(3, (2, 4, 6))
    print("标量左乘 (2 * v1):", 2 * v1)      # Vector(3, (2, 4, 6))
    print("向量逐元素乘:", v1 * v2)          # Vector(3, (4, 10, 18))
```
