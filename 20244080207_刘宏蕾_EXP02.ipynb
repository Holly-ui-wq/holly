```python
import numpy as np

print("="*20, "任务 1", "="*20)
# 1. 创建整数、浮点数、布尔值 ndarray
int_arr = np.array([1, 2, 3])
float_arr = np.array([1.0, 2.5, 3.1])
bool_arr = np.array([True, False, True])

for name, arr in [("整数", int_arr), ("浮点", float_arr), ("布尔", bool_arr)]:
    print(f"{name}数组: {arr}")
    print(f"  dtype: {arr.dtype}, shape: {arr.shape}, ndim: {arr.ndim}")

# 解释：
# NumPy 数组（ndarray）要求所有元素必须是相同的数据类型（同质），这使得它在内存中连续存储，运算效率极高。
# Python 列表可以包含任意不同类型的数据（异质），例如 [1, "a", True]，更灵活但运算效率较低。

print("\n" + "="*20, "任务 2", "="*20)
# 2. np.arange(12) 变形
arr_1d = np.arange(12)
print("原始一维数组:", arr_1d)

arr_3x4 = arr_1d.reshape(3, 4)
print("\n变形为 (3, 4):\n", arr_3x4)

arr_2x3x2 = arr_1d.reshape(2, 3, 2)
print("\n变形为 (2, 3, 2):\n", arr_2x3x2)

# 解释：
# 改变形状时，元素总数必须保持不变（原始元素个数 = 各维度大小的乘积）。
# 这里 12 = 3*4 = 2*3*2。

print("\n" + "="*20, "任务 3", "="*20)
# 3. 对 (3,4) 数组进行索引
# 假设第2行是指索引为 1 的行
row_2 = arr_3x4[1, :]
print("取第 2 行:", row_2, "形状:", row_2.shape)

# 取第 1 至第 3 列（索引 0 到 2，即切片 0:3）
col_1_to_3 = arr_3x4[:, 0:3]
print("\n取第 1 至第 3 列:\n", col_1_to_3, "\n形状:", col_1_to_3.shape)

# 取大于 6 的元素
greater_than_6 = arr_3x4[arr_3x4 > 6]
print("\n大于 6 的元素:", greater_than_6, "形状:", greater_than_6.shape)

print("\n" + "="*20, "任务 4", "="*20)
# 4. np.arange(10) 逻辑索引
arr_10 = np.arange(10)
print("原始数组:", arr_10)

# 使用 | (或)
cond1 = (arr_10 < 2) | (arr_10 > 7)
print("小于2或大于7 (使用 |):", arr_10[cond1])

# 使用 & (且)
cond2 = (arr_10 > 3) & (arr_10 < 7)
print("大于3且小于7 (使用 &):", arr_10[cond2])

# 使用 np.logical_or 和 np.logical_and
print("小于2或大于7 (使用 np.logical_or):", arr_10[np.logical_or(arr_10 < 2, arr_10 > 7)])
print("大于3且小于7 (使用 np.logical_and):", arr_10[np.logical_and(arr_10 > 3, arr_10 < 7)])

print("\n" + "="*20, "任务 5", "="*20)
# 5. 两个 (2, 2) 数组对比乘法
A = np.array([[1, 2], 
              [3, 4]])
B = np.array([[5, 6], 
              [7, 8]])

print("矩阵 A:\n", A)
print("矩阵 B:\n", B)

print("\n--- 逐元素乘法 ---")
print("A * B:\n", A * B)
print("np.multiply(A, B):\n", np.multiply(A, B))

print("\n--- 矩阵乘法 ---")
print("A @ B:\n", A @ B)
print("np.matmul(A, B):\n", np.matmul(A, B))

# 解释：
# * 和 np.multiply() 执行的是【逐元素乘法】（对应位置元素直接相乘），要求两个数组形状相同。
# @ 和 np.matmul() 执行的是【矩阵乘法】（行乘列求和），必须满足前一个矩阵的列数等于后一个矩阵的行数。
```

```python
import numpy as np
import pandas as pd

np.random.seed(42) # 设置随机种子以便复现结果（可选）

print("="*20, "任务 1", "="*20)
# 1. 生成 1000 个均值为 70，标准差为 10 的成绩，并限制在 0~100 之间
scores = np.random.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores, 0, 100)

print(f"样本数: {scores.size}")
print(f"最小值: {np.min(scores):.2f}")
print(f"最大值: {np.max(scores):.2f}")
print(f"均值: {np.mean(scores):.2f}")
print(f"标准差: {np.std(scores):.2f}")
print(f"极差: {np.ptp(scores):.2f}") # np.ptp 用于计算极差 (Peak to Peak)

print("\n" + "="*20, "任务 2", "="*20)
# 2. 从 1 到 10 中不放回抽取 5 个整数
population = np.arange(1, 11)

# 2.1 均匀抽样 (每个数字被抽到的概率相等)
uniform_sample = np.random.choice(population, size=5, replace=False)
print(f"均匀抽样结果: {uniform_sample}")

# 2.2 按整数大小成比例加权抽样 (权重大小即数字本身)
# 概率 p 必须和为 1，所以用每个数字除以总和
weights = population / np.sum(population) 
weighted_sample = np.random.choice(population, size=5, replace=False, p=weights)
print(f"加权抽样结果: {weighted_sample}")

print("\n" + "="*20, "任务 3", "="*20)
# 3. 随机排列与打乱
original_arr = np.arange(1, 11)
print(f"原数组: {original_arr}")

# 方式一：保留原数组 (np.random.permutation 返回一个新数组的副本)
permuted_arr = np.random.permutation(original_arr)
print(f"使用 permutation 后原数组: {original_arr}") # 没变
print(f"新生成的排列数组: {permuted_arr}")

# 方式二：原地打乱 (np.random.shuffle 直接在原数组上进行修改)
arr_to_shuffle = original_arr.copy() # 复制一份用于测试
np.random.shuffle(arr_to_shuffle)
print(f"使用 shuffle 后的数组: {arr_to_shuffle}") # 已经被改变了

print("\n" + "="*20, "任务 4", "="*20)
# 4. 缺失值处理
scores_with_nan = scores.copy()
# 将第 10、20、30 个元素（索引为 9, 19, 29）设为 np.nan
scores_with_nan[[9, 19, 29]] = np.nan

# 普通均值
mean_normal = np.mean(scores_with_nan)
# 忽略缺失值的均值
mean_ignore_nan = np.nanmean(scores_with_nan)
# 统计缺失值个数
nan_count = np.isnan(scores_with_nan).sum()

print(f"普通均值: {mean_normal}")
print(f"忽略缺失值的均值: {mean_ignore_nan:.2f}")
print(f"缺失值个数: {nan_count}")

print("\n" + "="*20, "任务 5", "="*20)
# 5. 将含有缺失值的模拟成绩及序号整理成表格变量
# 生成序号，假设从 1 开始编号
indices = np.arange(1, len(scores_with_nan) + 1)

# 使用 Pandas 创建 DataFrame (内存中的表格)
df_scores = pd.DataFrame({
    "序号": indices,
    "成绩": scores_with_nan
})

print("表格前 10 行数据：")
print(df_scores.head(10))
print("\n表格信息：")
df_scores.info()
```

```python
import numpy as np
import scipy.linalg as la
import scipy.optimize as opt
import scipy.stats as stats
import matplotlib.pyplot as plt

# 设置中文字体，防止图表中的中文显示为方块
plt.rcParams['font.sans-serif'] = ['SimHei']  # Windows使用黑体，Mac用户可改用 'Arial Unicode MS'
plt.rcParams['axes.unicode_minus'] = False    # 正常显示负号

print("="*20, "任务 1", "="*20)
# 1. 解线性方程组
# 3x + y = 9
# x + 2y = 8
A = np.array([[3, 1], 
              [1, 2]])
b = np.array([9, 8])

# 使用 scipy.linalg.solve 求解
solution = la.solve(A, b)
print(f"求解结果: x = {solution[0]:.2f}, y = {solution[1]:.2f}")

# 用 NumPy 验证 (A @ solution 是否等于 b)
verification = np.allclose(np.dot(A, solution), b)
print(f"NumPy 验证结果 (是否相等): {verification}")


print("\n" + "="*20, "任务 2", "="*20)
# 2. 使用 scipy.optimize.root 求方程 x^2 - 2 = 0 的根
def func2(x):
    return x**2 - 2

# 求正根 (初始猜测 1)
res_pos = opt.root(func2, x0=1)
print(f"【求正根】")
print(f"是否成功: {res_pos.success}")
print(f"求得正根: {res_pos.x[0]:.6f}")
print(f"函数值 (是否接近0): {func2(res_pos.x[0]):.2e}")

# 求负根 (初始猜测 -1)
res_neg = opt.root(func2, x0=-1)
print(f"\n【求负根】")
print(f"是否成功: {res_neg.success}")
print(f"求得负根: {res_neg.x[0]:.6f}")
print(f"函数值 (是否接近0): {func2(res_neg.x[0]):.2e}")


print("\n" + "="*20, "任务 3", "="*20)
# 3. 求 f(x) = x^2 + x + 2 的最小值
def func3(x):
    return x**2 + x + 2

# 初始猜测 0
res_min = opt.minimize(func3, x0=0)

# 解析解: 对于 ax^2+bx+c，极值点 x = -b/(2a) = -1/2 = -0.5，最小值 = (-0.5)^2 + (-0.5) + 2 = 1.75
print(f"优化算法求得的极小值点 x: {res_min.x[0]:.4f}")
print(f"优化算法求得的最小函数值: {res_min.fun:.4f}")
print(f"解析解极小值点 x: -0.5000, 解析解最小值: 1.7500")
print(f"与解析解是否一致: {np.allclose(res_min.x[0], -0.5)}")


print("\n" + "="*20, "任务 4", "="*20)
# 4. 独立样本 t 检验
np.random.seed(42)  # 固定随机种子

# 生成两组独立的模拟样本，各 100 个观测值
group1 = np.random.normal(loc=0, scale=1, size=100) # N(0, 1)
group2 = np.random.normal(loc=1, scale=1, size=100) # N(1, 1)

# 独立样本 t 检验
t_stat, p_val = stats.ttest_ind(group1, group2)

print(f"t 统计量: {t_stat:.4f}")
print(f"p 值: {p_val:.4e}")

# 判断：p 值小于 0.05 显著性水平，拒绝原假设（两组均值相等），说明两组样本有显著差异
if p_val < 0.05:
    print("判断：在 0.05 显著性水平下，p 值 < 0.05，拒绝原假设。两组样本均值存在显著差异。")
else:
    print("判断：在 0.05 显著性水平下，p 值 >= 0.05，无法拒绝原假设。两组样本均值无显著差异。")


print("\n" + "="*20, "任务 5", "="*20)
# 5. Matplotlib 可视化
# (1) 生成实验2中的模拟成绩数据（为了独立性，这里重新生成一次）
scores = np.random.normal(loc=70, scale=10, size=1000)
scores = np.clip(scores, 0, 100)

# 创建画布
fig = plt.figure(figsize=(14, 5))

# 子图1：直方图
ax1 = fig.add_subplot(1, 2, 1)
ax1.hist(scores, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
ax1.set_title('模拟考试成绩分布（直方图）')
ax1.set_xlabel('成绩')
ax1.set_ylabel('频数')

# 子图2：箱线图
ax2 = fig.add_subplot(1, 2, 2)
ax2.boxplot(scores, vert=True, patch_artist=True, boxprops=dict(facecolor='lightgreen'))
ax2.set_title('模拟考试成绩分布（箱线图）')
ax2.set_ylabel('成绩')

plt.tight_layout()
plt.show()

# (2) 绘制折线图 y = x, y = x^2, y = x^3 在 0 <= x <= 5 的区间
x = np.linspace(0, 5, 100)
y1 = x
y2 = x**2
y3 = x**3

plt.figure(figsize=(8, 5))
plt.plot(x, y1, label='y = x', color='blue', linestyle='-')
plt.plot(x, y2, label='y = x^2', color='green', linestyle='--')
plt.plot(x, y3, label='y = x^3', color='red', linestyle='-.')

plt.title('函数折线图 (0 <= x <= 5)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()  # 显示图例
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

```python
import pandas as pd
import numpy as np

# 设置 pandas 显示选项，方便查看
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# 准备初始数据
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82]
}

print("="*20, "任务 1", "="*20)
# 1. 创建 DataFrame，行索引为姓名；计算 Total 和 Average
df = pd.DataFrame(records).set_index("Name")

# 计算总分和平均分（只对三科成绩计算）
df['Total'] = df[['Math', 'English', 'Python']].sum(axis=1)
df['Average'] = df[['Math', 'English', 'Python']].mean(axis=1)

print("行索引:", df.index.tolist())
print("列名:", df.columns.tolist())
print("\n前 3 行数据:")
print(df.head(3))
print("\n基本统计信息:")
print(df.describe())


print("\n" + "="*20, "任务 2", "="*20)
# 2. 读取 Carol 的记录，并筛选平均分 >= 80 的学生
# 使用 loc 按标签(姓名)读取
print("使用 loc 读取 Carol:")
print(df.loc['Carol'])

# 使用 iloc 按位置读取 (Carol 在第 3 行，索引为 2)
print("\n使用 iloc 读取 Carol:")
print(df.iloc[2])

# 筛选平均分不低于 80 的学生
print("\n平均分 >= 80 的学生:")
print(df[df['Average'] >= 80])


print("\n" + "="*20, "任务 3", "="*20)
# 3. 计算列均值和行均值
subjects = ['Math', 'English', 'Python']

# axis=0 计算列均值（即各科平均分）
print("三科成绩列均值 (axis=0):")
print(df[subjects].mean(axis=0))

# axis=1 计算行均值（即每名学生的平均分）
print("\n每名学生的行均值 (axis=1):")
print(df[subjects].mean(axis=1))


print("\n" + "="*20, "任务 4", "="*20)
# 4. apply, agg, rank 的应用
# 使用 apply 计算每名学生的三科最高分
df['Max_Score'] = df[subjects].apply(np.max, axis=1)
print("每名学生的三科最高分:")
print(df['Max_Score'])

# 使用 agg() 一次得到每科的最小值、均值和最大值
print("\n各科成绩的极值与均值:")
print(df[subjects].agg(['min', 'mean', 'max']))

# 任选一列使用 rank() 排名 (以 Total 总分降序排名为例)
df['Rank'] = df['Total'].rank(ascending=False, method='min')
print("\n按总分降序排名:")
print(df[['Total', 'Rank']])


print("\n" + "="*20, "任务 5", "="*20)
# 5. 按 Group 分组统计，并使用 transform
# 统计每组人数和三科平均分
group_stats = df.groupby('Group').agg(
    Count=('Group', 'count'),
    Math_Avg=('Math', 'mean'),
    English_Avg=('English', 'mean'),
    Python_Avg=('Python', 'mean')
)
print("各组人数及三科平均分:")
print(group_stats)

# 用 transform('mean') 为每行添加所在组的 Python 平均分
df['Group_Python_Mean'] = df.groupby('Group')['Python'].transform('mean')
print("\n添加组均值后的数据:")
print(df[['Group', 'Python', 'Group_Python_Mean']])


print("\n" + "="*20, "任务 6", "="*20)
# 6. 用 filter() 保留 Python 平均分不低于 80 的组
# filter 接收一个作用于分组后子 DataFrame 的函数，返回布尔值
filtered_df = df.groupby('Group').filter(lambda x: x['Python'].mean() >= 80)
print("保留的组和学生 (Python 组均分 >= 80):")
print(filtered_df[['Group', 'Python']])


print("\n" + "="*20, "任务 7", "="*20)
# 7. 协方差矩阵与皮尔逊相关系数矩阵
cov_matrix = df[subjects].cov()
corr_matrix = df[subjects].corr()

print("三科成绩的协方差矩阵:")
print(cov_matrix)

print("\n三科成绩的皮尔逊相关系数矩阵:")
print(corr_matrix)
```
